<h1>در این تمرین از کتابخانه های آماده برای بارگذاری تصاویر استفاده نشده</h1>

<p>در این قسمت عکس ها را پس از آپلود کردن برای برگرداندن به حالت اصلی 90 درجه می‌چرخانیم و ذخیره می‌کنیم</p>

```python
import numpy as np
from imageio import imread, imwrite

def fix_rotation(img):
    return np.rot90(img, k=-1)  # rotate 90 degrees clockwise

# Load and fix image
img = imread("myimg.jpg")
img = fix_rotation(img)
imwrite("output_color.jpg", img)

# Convert to grayscale
if img.ndim == 3:
    gray = np.dot(img[..., :3], [0.299, 0.587, 0.114])
else:
    gray = img

gray_uint8 = np.clip(gray, 0, 255).astype(np.uint8)
imwrite("output_gray.jpg", gray_uint8)

# Black & White
threshold = 128
bw = (gray > threshold).astype(np.uint8) * 255
imwrite("output_bw.jpg", bw)
```

<p>در این مرحله برای افزایش کنتراست در عددی بزرگتر از 1 و برای کاهش در کمتر از 1 ضرب می‌کنیم</p>

```python
def adjust_contrast(img, factor):
    m = np.median(img)
    adjusted = (img - m) * factor + m
    return np.clip(adjusted, 0, 255).astype(np.uint8)

color_contrast_up = adjust_contrast(img, 1.5)
color_contrast_down = adjust_contrast(img, 0.5)
imwrite("output_color_contrast_up.jpg", color_contrast_up)
imwrite("output_color_contrast_down.jpg", color_contrast_down)

gray_contrast_up = adjust_contrast(gray, 1.5)
gray_contrast_down = adjust_contrast(gray, 0.5)
imwrite("output_gray_contrast_up.jpg", gray_contrast_up)
imwrite("output_gray_contrast_down.jpg", gray_contrast_down)
```

```python
gray_plus20 = np.clip(gray + 20, 0, 255).astype(np.uint8)
imwrite("output_gray_plus20.jpg", gray_plus20)
```

<p>در این بخش با احتمال یک درصد پیکسل را سفید یا سیاه می‌کنیم که همان نمک و فلفل است</p>

```python
def add_salt_pepper_noise(img, amount=0.02):
    noisy = img.copy()
    total_pixels = img.shape[0] * img.shape[1]
    num_salt = int(total_pixels * amount / 2)
    num_pepper = int(total_pixels * amount / 2)

    coords = [np.random.randint(0, i, num_salt) for i in img.shape[:2]]
    noisy[coords[0], coords[1]] = 255

    coords = [np.random.randint(0, i, num_pepper) for i in img.shape[:2]]
    noisy[coords[0], coords[1]] = 0

    return noisy.astype(np.uint8)

noisy_img = add_salt_pepper_noise(img, amount=0.02)
imwrite("output_color_noisy_saltpepper.jpg", noisy_img)
```
