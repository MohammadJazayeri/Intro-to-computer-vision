```python
from imageio import imread
import os

img = imread("myimg.jpg")

filename = "myimg.jpg"

with open(filename, 'rb') as f:
    header = f.read(54)
    print("Header (first 54 bytes):", header)

size = os.path.getsize(filename)




print("Shape (Resolution):", img.shape)  # (height, width, 3)
print("Data type:", img.dtype)
print("Min/Max pixel value:", img.min(), img.max())
print(f"\nFile Size: {size / 1024:.2f} KB")
```

\xff\xd8 tells us that this is a JPEG!

\xff\xe1 contains Exif metadata

\xaa\x98 is the length(size) of the metadata in our case 43672 bytes long.

Exif\x00\x00	Confirms Exif format

II	Little-endian (Intel)

\x2A	Identifies TIFF format

the image is JPEG and hence it uses Discrete Cosine Transformation(DCT) which is lossy.
